### Initialize Spark

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
        .appName('VatsanSparkApp') \
        .getOrCreate()
```

### Read and Show Data

```python
data = [
    (1, 'Alice'),
    (2, 'Bob')
]
columns = ['id', 'name']
csv_df = spark.read.csv('data.csv', header=True, inferSchema=True)
csv_df.show()

df = spark.createDataFrame(data, columns)
df.show()
```

### Select & Filter & Add & Renaming

```python
from pyspark.sql.functions import col, lit
# Select
select_df = csv_df.select(col('name'))
select_df.show()

# filter
filter_df = csv_df.filter(col('age') < 28)
filter_df.show()
filter_df = csv_df.filter(col('profession') == 'cybersecurity')
filter_df.show()

# add column
add_df = csv_df.withColumn('gender', lit('M'))
add_df.show()
add_df = csv_df.withColumn('age2', col('age')*2)
add_df.show()

# Renamed
ren_df = add_df.withColumnRenamed('age2', 'age_twice')
ren_df.show()
```
