### Initialize Spark

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
        .appName('VatsanSparkApp') \
        .getOrCreate()
```

### Read and Show Data

```python
data = [
    (1, 'Alice'),
    (2, 'Bob')
]
columns = ['id', 'name']
csv_df = spark.read.csv('data.csv', header=True, inferSchema=True)
csv_df.show()

df = spark.createDataFrame(data, columns)
df.show()
```
