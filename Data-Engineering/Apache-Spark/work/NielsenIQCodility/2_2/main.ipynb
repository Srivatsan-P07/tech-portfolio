```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('2_2').getOrCreate()
```

```python
prices = spark.read \
                .option('header', True) \
                .option('inferSchema', True) \
                .csv('prices.csv')
```

```python
from pyspark.sql.functions import to_date, date_format, avg, round, col, max
prices_with_month = prices.withColumn(
    'yyyy_mm', date_format(to_date('date', 'yyyy-MM-dd'), 'yyyy-MM')
)
```

```python
prices_with_month.groupBy('ticker', 'yyyy_mm') \
    .agg(
        round(avg(col('close')), 2).alias('avg_close'),
        max(col('volume')).alias('max_volume')
    ).orderBy('ticker', 'yyyy_mm').show()
```
