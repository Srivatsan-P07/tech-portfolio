```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('2_1').getOrCreate()
```

```python
customers = spark.read \
                .option('header', True) \
                .option('inferSchema', True) \
                .csv('customers.csv')

transactions = spark.read \
                    .option('header', True) \
                    .option('inferSchema', True) \
                    .csv('transactions.csv')
```

```python
from pyspark.sql.functions import col, sum
transactions_2023 = transactions.filter(col('txn_date') > '2023-01-01')

cust_tran = customers.join(transactions, ['customer_id'], how='inner')
grouped_data = cust_tran.groupBy('customer_id', 'name', 'country').agg(sum(col('amount')).alias('total_spent'))
grouped_data.filter(col('total_spent')>= 500).show()
```
