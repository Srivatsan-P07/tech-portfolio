```python
import functools

def my_decorator_function(func):
    @functools.wraps(func)
    def wrapper():
        print("before function")
        func()
        print("after function")
    return wrapper

@my_decorator_function
def custom_func():
    print("hello from custom func")

@my_decorator_function
def custom_func2():
    print("hello from custom func2")

```
