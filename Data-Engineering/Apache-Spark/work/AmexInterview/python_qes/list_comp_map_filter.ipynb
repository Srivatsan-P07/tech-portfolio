```python
nums = [1,2,3,4,5]

# List Comprehension, single line execution

num_list = [x*x for x in nums if x%2 == 0]
print(num_list)
```

```python
# map and filter
# map(lambda function, values)
# filter(lambda function, values)
# here , map(lambda function, (filter(lambda function, values))
num_list = list(
    map(lambda x: x*x,
        filter(lambda x: x%2==0, nums))
)
print(num_list)
```
