```python
# Broadcast join , is a readonly variable
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('broadcast_variable').getOrCreate()
sc = spark.sparkContext

USER_DATA = [('john', 'US'), ('bob', 'INDIA')]

country_codes = {
    'US': 'UNITED STATES',
    'INDIA': 'BHARAT'
}

broadcast_var = sc.broadcast(country_codes)

rdd = sc.parallelize(USER_DATA)

# list of tuple for rdd and dataframes
output = rdd.map(lambda x: [
    x[0], broadcast_var.value.get(x[1])
])
output2 = output.collect()
print(output2)
df = spark.createDataFrame(output2, ['username', 'nation'])
df.show()
```

```python
# Accumulators
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('Accumulator').getOrCreate()
sc = spark.sparkContext

user_data = [10, -20, 30, -40, 50]

accumulator_counter = sc.accumulator(0)

rdd = sc.parallelize(user_data)

def rdd_process(x):
    global accumulator_counter
    if x<0:
        accumulator_counter.add(1)
        return None
    else:
        return x

output = [x for x in rdd.map(lambda x: rdd_process(x)).collect() if x is not None]
print(output)
```
