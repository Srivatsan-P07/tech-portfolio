```python
# Catalyst is Spark SQL’s query optimization engine. 
# It transforms a user query into an efficient execution plan through multiple stages:

from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('CatalystOptimizer').getOrCreate()

data = [
    ("Alice", 60000),
    ("Bob", 40000),
    ("Charlie", 70000)
]

df = spark.createDataFrame(data, ['name', 'salary'])
filtered_df = df.filter(df.salary > 50000).select('name')

filtered_df.show()
```

```python
filtered_df.explain(True)
```
