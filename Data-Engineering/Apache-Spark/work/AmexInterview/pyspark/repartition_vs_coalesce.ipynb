```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Repartition").getOrCreate()

sc = spark.sparkContext
sc.setJobGroup("repartition", 'repartition')

data = [
    (1, "A"),
    (2, "B"),
    (3, "C"),
    (4, "D"),
    (5, "E"),
    (6, "F")
]

df = spark.createDataFrame(data, ['id', 'value'])
df.show()
```

```python
num_part = df.rdd.getNumPartitions()
print(num_part)

# repartition
df = df.repartition(5)
df.write.mode("overwrite").option('header', 'true').csv("output_repartition")

df = df.coalesce(2)
df.write.mode('overwrite').option('header', 'true').csv("output_coalesce")
```
