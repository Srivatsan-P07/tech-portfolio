```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Repartition").getOrCreate()

sc = spark.sparkContext
sc.setJobGroup("repartition", 'repartition')

data = [
    (1, "A"),
    (2, "B"),
    (3, "C"),
    (4, "D"),
    (5, "E"),
    (6, "F")
]

df = spark.createDataFrame(data, ['id', 'value'])
df.show()
```

```python
num_part = df.rdd.getNumPartitions()
print(num_part)

# repartition if you want to repartition after a filter for distributed process
df = df.repartition(5)

# Coalesce if you want to write as output and dont want many files.
df.coalesce(1).write.mode("overwrite").csv("output/test")
```
