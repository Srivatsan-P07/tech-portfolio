```python
data = [
    ("apple", 15),
    ("apple", 30)
]

from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('reduceByKey').getOrCreate()

sc = spark.sparkContext

data = sc.parallelize(data)

result = data.reduceByKey(lambda x,y: x+y)
print(result)
map_result = data.map(lambda x: x[1]*(-1))
map_output = map_result.collect()
print(map_output)

output = result.collect()

print(output)
```
